```python
transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor()
])
```

```python
class resnet_feature_extractor(torch.nn.Module):
    def __init__(self):
        """This class extracts the feature maps from a pretrained Resnet model."""
        super(resnet_feature_extractor, self).__init__()
        self.model = resnet50(weights=ResNet50_Weights.DEFAULT)

        self.model.eval()
        for param in self.model.parameters():
            param.requires_grad = False



        # Hook to extract feature maps
        def hook(module, input, output) -> None:
            """This hook saves the extracted feature map on self.featured."""
            self.features.append(output)

        self.model.layer2[-1].register_forward_hook(hook)
        self.model.layer3[-1].register_forward_hook(hook)

    def forward(self, input):

        self.features = []
        with torch.no_grad():
            _ = self.model(input)

        self.avg = torch.nn.AvgPool2d(3, stride=1)
        fmap_size = self.features[0].shape[-2]         # Feature map sizes h, w
        self.resize = torch.nn.AdaptiveAvgPool2d(fmap_size)

        resized_maps = [self.resize(self.avg(fmap)) for fmap in self.features]
        patch = torch.cat(resized_maps, 1)            # Merge the resized feature maps
        patch = patch.reshape(patch.shape[1], -1).T   # Craete a column tensor

        return patch
```

```python
# imports (safe)
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
from tqdm import tqdm
import math

# PyG imports
from torch_geometric.data import Data
from torch_geometric.nn import GATConv

# image resize
import torch.nn.functional as F

# device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print("DEVICE:", device)
```

```python
def build_patch_graph_from_features(patch_features, grid_h=None, grid_w=None, add_positional=True):
    """
    Convert patch_features (tensor [P, F]) into a PyG Data with grid adjacency.
    If grid_h/w not provided, try to infer a square grid (sqrt(P)).
    Returns: Data(x=node_features, edge_index)
    - adds optional positional coords as extra dims if add_positional=True
    """
    # patch_features: torch.Tensor [P, F]
    if not isinstance(patch_features, torch.Tensor):
        patch_features = torch.tensor(patch_features, dtype=torch.float)

    P, F = patch_features.shape

    if grid_h is None or grid_w is None:
        sq = int(math.sqrt(P))
        if sq * sq == P:
            grid_h = grid_w = sq
        else:
            # fallback: try factors. But default to P x 1 if cannot infer
            grid_h = P
            grid_w = 1

    # Node features start with the patch feature vector
    x = patch_features.clone().float()

    if add_positional:
        # add normalized (y,x) grid coordinates as extra two dims
        coords = []
        for idx in range(P):
            r = idx // grid_w
            c = idx % grid_w
            coords.append([r / (grid_h - 1) if grid_h>1 else 0.0, c / (grid_w - 1) if grid_w>1 else 0.0])
        coords = torch.tensor(coords, dtype=torch.float)
        x = torch.cat([x, coords], dim=1)  # new dim F+2

    # Build 4-neighbour grid adjacency (or 8-neighbour if you want)
    edge_list = []
    for r in range(grid_h):
        for c in range(grid_w):
            i = r * grid_w + c
            # right neighbour
            if c + 1 < grid_w:
                j = r * grid_w + (c+1)
                edge_list.append([i, j]); edge_list.append([j, i])
            # down neighbour
            if r + 1 < grid_h:
                j = (r+1) * grid_w + c
                edge_list.append([i, j]); edge_list.append([j, i])
            # (optional) diagonal neighbours - comment out if not wanted
            # if r+1 < grid_h and c+1 < grid_w:
            #     j = (r+1)*grid_w + (c+1)
            #     edge_list.append([i,j]); edge_list.append([j,i])

    if len(edge_list) == 0:
        edge_index = torch.empty((2,0), dtype=torch.long)
    else:
        edge_index = torch.tensor(edge_list, dtype=torch.long).t().contiguous()

    data = Data(x=x, edge_index=edge_index)
    return data, (grid_h, grid_w)
```

```python
class GATAutoEncoder(nn.Module):
    def __init__(self, in_dim, hidden_dim=256, latent_dim=128, heads=4, dropout=0.2):
        super().__init__()
        self.encoder1 = GATConv(in_dim, hidden_dim // heads, heads=heads, concat=True, dropout=dropout)
        self.encoder2 = GATConv(hidden_dim, latent_dim // heads, heads=heads, concat=True, dropout=dropout)

        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, hidden_dim),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, hidden_dim // 2),
            nn.ReLU(),
            nn.Linear(hidden_dim // 2, in_dim)
        )

    def forward(self, x, edge_index):
        h = torch.relu(self.encoder1(x, edge_index))
        h = torch.relu(self.encoder2(h, edge_index))
        recon = self.decoder(h)
        return recon, h
```

```python
class GATAutoencoder(nn.Module):
    def __init__(self, in_channels, hidden_channels=128, dropout=0.2):
        super(GATAutoencoder, self).__init__()
        self.encoder = GATConv(in_channels, hidden_channels, heads=2, concat=True)
        self.decoder = nn.Sequential(
            nn.Linear(hidden_channels * 2, in_channels),
            nn.ReLU(),
            nn.Linear(in_channels, in_channels)
        )
        self.dropout = dropout

    def forward(self, x, edge_index):
        # Encoder
        x = F.elu(self.encoder(x, edge_index))
        x = F.dropout(x, p=self.dropout, training=self.training)
        # Decoder
        x_recon = self.decoder(x)
        return x_recon

def train_gat_autoencoder_on_normal_images(
    backbone_model,
    train_folder,
    model_save_path='gat_ae.pth',
    epochs=10,
    lr=5e-4,
    batch_size=8,
    grid_size=None,
    print_every=1
):
    backbone_model.eval()

    img_paths = list(Path(train_folder).glob('*.*'))
    if len(img_paths) == 0:
        raise ValueError(f"No images found in {train_folder}")

    # --- Build model shape from first image ---
    with torch.no_grad():
        img = Image.open(img_paths[0]).convert('RGB')
        inp = transform(img).unsqueeze(0).to(device)
        patch = backbone_model(inp)
        P, feat_dim = patch.shape

    in_dim = feat_dim + 2
    #gat_model = GATAutoEncoder(in_dim=in_dim, hidden_dim=256, latent_dim=128, dropout=0.2).to(device)
    gat_model = GATAutoencoder(in_channels=1538, hidden_channels=64).to(device)


    optimizer = optim.Adam(gat_model.parameters(), lr=lr, weight_decay=1e-5)
    mse_loss = nn.MSELoss()
    cosine = nn.CosineSimilarity(dim=1)

    for ep in range(epochs):
        epoch_loss = 0.0
        pbar = tqdm(img_paths, desc=f"Epoch {ep+1}/{epochs}")
        for p in pbar:
            img = Image.open(p).convert('RGB')
            inp = transform(img).unsqueeze(0).to(device)
            with torch.no_grad():
                patch = backbone_model(inp)

            data, (gh, gw) = build_patch_graph_from_features(
                patch.cpu(),
                grid_h=grid_size[0] if grid_size else None,
                grid_w=grid_size[1] if grid_size else None,
                add_positional=True
            )
            data.x = data.x.to(device)
            data.edge_index = data.edge_index.to(device)
            # Add Gaussian noise to input features (augmentation)


            # ---------- forward + loss ----------
            recon = gat_model(data.x, data.edge_index)

            feat_dim_local = patch.shape[1]
            recon_feat = recon[:, :feat_dim_local]
            target_feat = data.x[:, :feat_dim_local]

            mse = mse_loss(recon_feat, target_feat)
            cos_sim = torch.mean(1 - cosine(recon_feat, target_feat))
            loss = mse + 0.5 * cos_sim

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            epoch_loss += loss.item()
            pbar.set_postfix(loss=epoch_loss / (pbar.n if pbar.n > 0 else 1))

        if (ep + 1) % print_every == 0:
            print(f"Epoch {ep+1}/{epochs} avg loss: {epoch_loss/len(img_paths):.6f}")

    torch.save(gat_model.state_dict(), model_save_path)
    print("✅ Saved improved GAT-AE ->", model_save_path)
    return gat_model

    # Add Gaussian noise to input features (augmentation)
    noise = torch.randn_like(data.x) * 0.02  # small standard deviation
    noisy_input = data.x + noise
    recon = gat_model(noisy_input.to(device), data.edge_index)


# Usage example (do not overwrite backbone or memory_bank)
# gat_model = train_gat_autoencoder_on_normal_images(backbone, 'carpet/train/good', epochs=6)
```

```python
def gat_ae_infer_image_heatmap(image_pil, backbone_model, gat_model, up_size=(224,224)):
    """
    image_pil: PIL.Image
    returns: segm_map_upsampled (H,W) as numpy array, per_patch_scores (P,)
    """
    backbone_model.eval()
    gat_model.eval()

    inp = transform(image_pil).unsqueeze(0).to(device)
    with torch.no_grad():
        patch = backbone_model(inp)  # [P, feat_dim]
    # build graph
    data, (gh, gw) = build_patch_graph_from_features(patch.cpu(), grid_h=None, grid_w=None, add_positional=True)
    data.x = data.x.to(device)
    data.edge_index = data.edge_index.to(device)

    with torch.no_grad():
        recon = gat_model(data.x, data.edge_index)
    # compute per-node MSE between recon_feat and original feat
    feat_dim = patch.shape[1]
    recon_feat = recon[:, :feat_dim]
    orig_feat = data.x[:, :feat_dim]
    # per-node MSE
    per_node_mse = torch.mean((recon_feat - orig_feat)**2, dim=1)  # [P,]
    per_node_mse = per_node_mse.cpu().numpy()

    # reshape to grid and upsample to image size
    P = per_node_mse.shape[0]
    # infer grid dims
    sq = int(math.sqrt(P))
    if sq*sq == P:
        gh = gw = sq
    else:
        # fallback: try to use previously returned gh, gw
        gh, gw = gh, gw

    segm_map = per_node_mse.reshape(1, 1, gh, gw).astype(np.float32)  # 1x1xhxw
    segm_map_t = torch.tensor(segm_map)
    segm_up = F.interpolate(segm_map_t, size=up_size, mode='bilinear', align_corners=False)
    segm_up = segm_up.squeeze().cpu().numpy()
    # normalize to 0..1
    segm_up = (segm_up - segm_up.min()) / (segm_up.max() - segm_up.min() + 1e-8)

    return segm_up, per_node_mse
```

```python
# 📊 Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score, confusion_matrix, ConfusionMatrixDisplay

# Example labels — replace with your actual test labels & predictions
class_names = ['good', 'cut', 'hole', 'color', 'thread']
y_true = ['good','cut','hole','color','hole','good','cut','thread','thread','color']
y_pred = ['good','cut','hole','color','good','good','cut','thread','thread','color']

# ✅ Compute per-class metrics
metrics_data = []
for cls in class_names:
    y_true_bin = [1 if y == cls else 0 for y in y_true]
    y_pred_bin = [1 if y == cls else 0 for y in y_pred]
    precision = precision_score(y_true_bin, y_pred_bin, zero_division=0)
    recall = recall_score(y_true_bin, y_pred_bin, zero_division=0)
    f1 = f1_score(y_true_bin, y_pred_bin, zero_division=0)
    acc = accuracy_score(y_true_bin, y_pred_bin)
    metrics_data.append([cls, precision, recall, f1, acc])

# 📋 Create a DataFrame
df_metrics = pd.DataFrame(metrics_data, columns=['Defect Type', 'Precision', 'Recall', 'F1-Score', 'Accuracy'])
df_metrics.set_index('Defect Type', inplace=True)
df_metrics = df_metrics.round(3)

# 🧷 Display styled table
styled = df_metrics.style.background_gradient(cmap='Blues').set_caption("📊 Per-Defect Metrics Summary")
display(styled)

# 📉 Bar Plot Visualization
plt.figure(figsize=(10,6))
df_metrics.plot(kind='bar', figsize=(10,6))
plt.title("Performance Metrics per Defect Type", fontsize=14, weight='bold')
plt.ylabel("Score")
plt.ylim(0,1.1)
plt.xticks(rotation=0)
plt.legend(loc='lower right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# 🔥 Confusion Matrix
cm = confusion_matrix(y_true, y_pred, labels=class_names)
plt.figure(figsize=(7,6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Purples', xticklabels=class_names, yticklabels=class_names)
plt.title("Confusion Matrix", fontsize=14, weight='bold')
plt.xlabel("Predicted Labels")
plt.ylabel("True Labels")
plt.show()
```
